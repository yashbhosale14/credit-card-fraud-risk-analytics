```python
def classify_time(hour):

    if hour < 4:
        return "00-04 Late Night"
    elif hour < 8:
        return "04-08 Early Morning"
    elif hour < 12:
        return "08-12 Morning"
    elif hour < 16:
        return "12-16 Afternoon"
    elif hour < 20:
        return "16-20 Evening"
    else:
        return "20-24 Night"

df["Time_Period"] = df["Hour"].apply(classify_time)
```

```python
df["Amount_Bucket"] = pd.cut(
    df["Amount"],
    bins=[-1, 10, 50, 100, 500, 1000, np.inf],
    labels=[
        "0-10",
        "10-50",
        "50-100",
        "100-500",
        "500-1000",
        "1000+"
    ]
)
```

# Fraud By Hour

```python
fraud_by_hour = (
    df.groupby("Hour")
      .agg(
          Total_Transactions=("Transaction_ID", "count"),
          Fraud_Transactions=("Class", "sum"),
          Total_Amount=("Amount", "sum"),
          Fraud_Amount=("Amount", lambda x: x[df.loc[x.index, "Class"] == 1].sum())
      )
      .reset_index()
)
```

```python
fraud_by_hour["Fraud_Rate"] = (
    fraud_by_hour["Fraud_Transactions"]
    / fraud_by_hour["Total_Transactions"]
    * 100
)
```

```python
fraud_by_hour.sort_values(
    "Fraud_Rate",
    ascending=False
)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 5))

sns.lineplot(
    data=fraud_by_hour,
    x="Hour",
    y="Fraud_Rate",
    marker="o"
)

plt.title("Historical Fraud Rate by Hour")
plt.xlabel("Hour of Day")
plt.ylabel("Fraud Rate (%)")
plt.xticks(range(24))
plt.grid(alpha=0.3)

plt.show()
```

# Fraud By Amount

```python
fraud_by_amount = (
    df.groupby("Amount_Bucket", observed=True)
      .agg(
          Total_Transactions=("Transaction_ID", "count"),
          Fraud_Transactions=("Class", "sum"),
          Total_Amount=("Amount", "sum")
      )
      .reset_index()
)
```

```python
fraud_by_amount["Fraud_Rate"] = (
    fraud_by_amount["Fraud_Transactions"]
    / fraud_by_amount["Total_Transactions"]
    * 100
)
```

# Analyze V1–V28

```python
feature_summary = []

for col in [f"V{i}" for i in range(1, 29)]:

    legitimate_mean = df.loc[
        df["Class"] == 0, col
    ].mean()

    fraud_mean = df.loc[
        df["Class"] == 1, col
    ].mean()

    feature_summary.append({
        "Feature": col,
        "Legitimate_Mean": legitimate_mean,
        "Fraud_Mean": fraud_mean,
        "Absolute_Mean_Difference": abs(
            fraud_mean - legitimate_mean
        )
    })

feature_summary = pd.DataFrame(feature_summary)

feature_summary.sort_values(
    "Absolute_Mean_Difference",
    ascending=False
)
```
